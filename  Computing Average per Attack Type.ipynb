```python
import pandas as pd
import numpy as np

# Load the CSV file
df = pd.read_csv("attack_execution_time_log.csv")

# Convert to correct types if needed
df["Execution_Time"] = df["Execution_Time"].astype(float)



# Group by Attack_Type
grouped = df.groupby("Attack_Type")["Execution_Time"]

# Print statistics
print("📊 Execution Time Summary per Attack Type:")
for label, times in grouped:
    print(f"{label}:")
    print(f"  Count:      {len(times)}")
    print(f"  Avg Time:   {np.mean(times):.6f}s")
    print(f"  Max Time:   {np.max(times):.6f}s")
    print(f"  Min Time:   {np.min(times):.6f}s")
    print()
# 🔽 Total Average Across All Attacks
overall_avg = df["Execution_Time"].mean()
print(f"🧮 Total Average Execution Time (all attacks): {overall_avg:.6f}s")
```

```python
import pandas as pd
import numpy as np

# Load the updated CSV file
df = pd.read_csv("attack_execution_time_log.csv")

# Convert relevant columns to float
df["Model_Prediction_Time"] = df["Model_Prediction_Time"].astype(float)
df["Explanation_Time"] = df["Explanation_Time"].astype(float)
df["Control_to_End_Time"] = df["Control_to_End_Time"].astype(float)
df["Total_Time"] = df["Total_Time"].astype(float)

# Group by Attack_Type
grouped = df.groupby("Attack_Type")

# Print statistics per attack type
print("📊 Execution Time Breakdown per Attack Type:")
for label, group in grouped:
    print(f"{label}:")
    print(f"  Count:                  {len(group)}")
    print(f"  Avg Model Prediction:   {group['Model_Prediction_Time'].mean():.6f}s")
    print(f"  Avg Explanation Time:   {group['Explanation_Time'].mean():.6f}s")
    print(f"  Avg Control to End:     {group['Control_to_End_Time'].mean():.6f}s")
    print(f"  Avg Total Time:         {group['Total_Time'].mean():.6f}s")
    print()

# 🔽 Overall Averages
print("🧮 Overall Average Times Across All Attacks:")
print(f"  Avg Model Prediction:   {df['Model_Prediction_Time'].mean():.6f}s")
print(f"  Avg Explanation Time:   {df['Explanation_Time'].mean():.6f}s")
print(f"  Avg Control to End:     {df['Control_to_End_Time'].mean():.6f}s")
print(f"  Avg Total Time:         {df['Total_Time'].mean():.6f}s")
```
